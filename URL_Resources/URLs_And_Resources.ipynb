## **Navigating the Internet's Resources**

Messy means complexity. Just think about how you'd find your way in big city without street name!

The Internet is pretty much a giant city of electronic information and URLs are the standardized way used to uniquely locate Internet's resources.
URL stands for Uniform Resource Locator. URLs are part of a broader resource identifier called URI (Uniform resource identifier) which includes in addition URN (Uniform resource name).

URLs locate resources by telling you where they are located while URNs identify them by using their names. These might appear the same but they are really different. Just like you might move from one place to another, resources on the web can also change locations. If a friend doesn't know that and you've moved and tries to surprise at your old address, they definitely won't find you. This is one of the URLs drawbacks that URNs are trying to solve. 

However URLs are the most common way to locate Internet's resouces. They are used by browser and applications to find, use and share data resources on the web.

Say you want to fetch the URL in the figure below (from HTTP the definitive guide book p24):

- The first part of the URL (http) is the URL scheme. The scheme tells a web client how to access the resource. In this case, the URL says to use the HTTP protocol.
  
- The second part of the URL (www.joes-hardware.com) is the server location. 
This tells the web client where the resource is hosted.

- The third part of the URL (/seasonal/index-fall.html) is the resource path. The path tells what particular resource on the web server is being requested.

## **URL syntax**

The syntax of URL depends on the URL scheme or the protocol used to communicate. Most URL schemes base their URL syntax on this nine-part general format:

\<scheme>://\<user>:\<password>@\<host>:\<port>/\<path>;\<params>?\<query>#\<frag>

Almost no URLs contains all these components. But the most three parts of a URL are the *scheme*, the *host* and the *path*.

- **Scheme**:
  
  scheme refers to the protocol ( we'll dive into protocol in the next chapter) that the applications use the URL need to speak.
  In our case, it is the HTTP protocol.

- **Hosts and Ports**:
  
  The host or hostname is the textual domain name, more understandable by human, that identify which machine on the internet hosts the resouces you're requesting. This hostname, when provided to a brwoser, is converted into IP address. The port component defines the network port on which the server is listening.

- **Usernames and Passwords:**

    FTP (file transport protocol) servers are the most frequent ones that require username and a password. If an application uses a URL scheme that requires username and password, such as FTP, it generally will insert a default username and password if they are not supplied (for example: ftp://ftp.prep.ai.mit.edu/pub/gnu).

- **Paths:**

    The path locates the resources you're resquesting on the server's filesystem. In the example above, the path is /seasonal/index-fall.html. It resembles of a hierarchical path on a Unix filesystem. Each segment path can have its own params component.

- **Parameters:**

    Parameters provide any additional information that applications need to communicate correctly with servers. They are name/value pairs seperated from the rest of the URL and to each other by the ";" character. As we mention just above, each path segment can have its own params component (http://www.joes-hardware.com/hammers;sale=false/index.html;graphics=true).

- **Query String:**

     Queries are relevant for database services. They can be asked question or queries to narrow down the resources being requested. By convention, they are formatted as a series of "name=value" pairs separated the "&" character.
  
- **Fragments:**

  Fragments might be the best solution for you when you want to access to a particular section of a large file. Fragments are only related to clients, servers work with the whole resources. When you provide a URL with fragment to your browser, it doesn't pass the fragment along to server. Once it get back the entire resources, it then uses the fragment to diplay the part of the resource in which you're interested.
