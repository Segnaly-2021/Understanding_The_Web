# **HTTP Messages**

Welcome in the second notebook of our series on how the web works!

In this one, we'll dive into HTTP messages.

## **Overview:**

HTTP is the Hypertext Transport Protocol, a language used by applications to exchange data across the Internet. You can think of HTTP messages as packages or envelops in which HTTP programs wrap data being transmitted. As we saw in the previous chapter, web contents are hosted on web servers waiting quitely for you to being requested. For example when you open a web page in your browser, here's what happens behind the scenes:

- Your browser sends an HTTP messages to the server that hosts the specified page (we'll explore the initial connection process in the next chapter).
- The server receives and processes your request.
- Finally, it sends the requested page back to your browser, which then displays it to you.

As you might guess, there are two type of HTTP messages, depending on their origin and destination. Messages from clients to servers are called *request messages,* while those from servers back to clients are *response messages*.

Regardless of which type they belong to, HTTP messages are composed of three parts: the *start line*, the *headers* and the *optional entity body*.

- **The start line**: its syntax varies according to the type of message

    -  For request message: *\<HTTP method> \<URI> \<HTTP version>*

        The start line of a request message tells the server where to find the requested resources (URI), what action to take (HTTP method) and which version of HTTP the client is using.
       
    -  For response message: *\<HTTP version> \<Status code> \<Reason phrase>*

        The start line of a response message indicates us what happened during the processing of the request (status code),provides a reason phrase that explains the status code, and specifies the version of HTTP that was used.

- **The headers**:

  the header fields add additional information to request and response messages. They are basically just lists of name/value pairs, and usually describe the entity body. For instance the size and the type of the body can be  indicated by the *content-length* and the *content-type* header fields, respectively:
  
  - content-length: 25
  - content-type: application/json

  **Note**: application is a broad type that includes none traditional data type such as text and image.
   

- **The optional entity body:**

  The entity body contains the data being transmitted. It is optional because not all HTTP methods require data to be included, such as  GET, DELETE, OPTIONS, ...

## **HTTP Methods:**

HTTP methods start request messages as we saw in the overview section. They tell the servers what action to take.

There are several HTTP methods but the most common are:

- **GET:** is used to ask a server to send a resource.

- **POST:** sends data to a server for processing.

- **DELETE:** deletes a resource.

- **PUT:** is used to upload a resource on a server.

## **Status codes:**

Just as HTTP methods tell the server what to do, status code indicates what happened during the processing of the request.

They are classified into five broad groups, as indicated in the table below.

<figure style="text-align: center; padding: 20px;">
    <img src="./status_codes.png" style="width:500px;">
    <figcaption> Source: HTTP The Definitive Guide book, P49 </figcaption>
</figure>

Status codes between 200 and 299 represent success. Codes between 300 and 399 means that the resource has been moved. Codes between 400 and 499 indicate that the client did something wrong in the request. Codes between 500 and 599 signify something went wrong on the server.

As you can see, there are several status codes But you don't need to memorize them all. They are always associated with a reason phrase that describes them.

If you get to this point, that means everything went well and you got the 200 status code. Time to celebrate!
