## **Overview**

Websites are just precreated content hosted on web servers. Just like when you create a Word document and store it in your filesystem, developers create and store websites on web servers. The only difference is that everybody can access it. It just requires to have access on Internet and a web browser to open a website. By doing this, you are making request to the server. Web servers rely heavily on HTTP protocol, which itself relies on TCP/IP connections(see ... for more details) between a *client*(generally browser) and a *web server*.

When a client opens a website, many things happen behind the scenes. The web server sets up the connection, receives and processes the HTTP message, finds the resource(in this case, a web page), constructs the reponse, and sends it back to the client. 

Usually when we talk about web servers we think about those hidden giant machines that hold web content such as HTML pages or dynamic resources generated by softwares. But web server can be anything, from those giant machines or the cloud to your own computer. Even a single software can be seen as a web server. They come in all shape and size:

- **General-purpose software web server**

  These softwares can be run on a standard, network-enabled computer system.
  
  *Examples: Nginx, Apache, Uvicorn, Gunicorn*
  
   If Nginx and Apache are general-purpose software web servers, meaning that they can be used for a wide range of web apps(HTML, PHP,JavaScript, ...),    Uvicorn and Gunicorn are used only for python web app.

- **Web server Appliances**:

  Whenever you plan to buy a Web server and don't want to bother yourself with installing software web server and configurations, web server appliances   might be the best choice for you. Web server appliances are prepackaged software/hardware solutions for a specific task.

  *Example: Barracuda Web Application Firewall*

  The Barracuda Web Application Firewall is a hardware appliance designed to protect web applications from threats and attacks.

## **What Real Web Servers Do**

In this section, we will dive into what real web servers do. No matter what forms they take, web servers perform a set of actions:

### **Step 1: Accepting Client Connections**

Web servers seat idle, waiting for a connection request from clients. Before exchanging any data, servers need to establish connections with clients.

You can think of it as the greeting words you'd express to someone before asking for directions back in the days before Google Maps. Typically, you'd say "Hello!", wait until they respond with "Hi!", and then you proceed to ask your questions. 

This allows you, as well as the web server with its clients, to check, among other things, whether you and your interlocutor are speaking the same language. The web server spoken language is the HTTP protocol.



### **Step 2: Receiving and Processing Request Messages**

Once connections are open, servers and clients can send data to each other. Resquest messages are divided into small chuncks called packets and arrive to the server erractically. The web server needs to store temporarily these packets until it receives enough data to parse it and make sense of it.

The request messages are made up of three parts: 

- The **start line**: includes the *HTTP method*(see ... for more details), the *resource identifier*, and the *HTTP version*.
- The **headers**: are metadata to give information about the entity body(content-type, content-length, ...)
- The **entity body**: if one was generated; Not every request message has a body content.

High-performance web servers can handle simultaneously thousands of connections, by which they receive millions of request messages. Web servers adopt multiple architecture to face this growing demand:

- **Single-threaded web servers (fig a)**:

  This type of web server processes one request at a time until completion. They are simple to implement but can create serious performance problem.

- **Multiprocess and Multithreaded web servers (fig b)**:

  Processes and Threads are the way that Operating system handles multitasking. Basically a process is a program in execution while a thread is a    subprocess(a set of instructions in execution). Even if CPUs can run a single thread at a time, your operating system contains a component called  scheduler which defines two things:
  - which thread to push to the CPU, and
  - the frequency at which it switch to another thread.
  
  You can think of scheduler as a switch that turn on and off programs but it does it so fast that we, as human, think computers run multiple programs  at the same time.

  Because of this, web servers can watch for multiple connections simultaneously and associate a process to each of them.

- **Multiplexed I/O servers (fig c):**

  In this configuration, processes and threads are not tied up waiting  on idle connections. The server use an open connection list to watch for any changes(meaning an arriving data through a connection). If a connection changes, the server processes it and puts it back to the open connection list when the transaction completes.

- **Multiplexed multithreaded web servers (fig d)**:

  These servers combine  multiplexing and multithreading to take advantage of multiple CPUs in the worker pool.

#### **Step 3: Mapping and Accessing Resources**

Web servers just host preceated contents and deliver them whenever a client request them. Identifying resources in a web server is similar what you and I do with our own computer, which is navagating through the filesystem. Web servers filesystem contains a special folder called **document root** or **docroot** for web content.

When a client request for example an HTML page, the web server takes the specified URI from the start line and appends it to the document root to get the absolute path of that file.

Web servers can also host multiple websites each with its own docroot. In this case, the *Host header* will help distinguish websites.  



#### **Step 4: Building Responses**

Once the resources are identified, the web server performes the action described by the HTTP method and returns the response message. This response is made up of three parts: 

- **the status code**: that tells the client what happened
- **the response headers**: the WS is responsible of determining the type of the body content. The most common way is to use MIME-typing for the file extension
- **the response body**: if one was generated

#### **Step 5: Sending Responses**

Just like the request messages, response messages travel across the Internet inside packets. But this time, they are going from the server back to the client.

#### **Step 6: Closing Connections**

Depending on the type of the connection, whether it is persistent or not, the server will have to close it when the transaction is done.
Standard connections are closed as soon as the transaction they were carrying is complete.
