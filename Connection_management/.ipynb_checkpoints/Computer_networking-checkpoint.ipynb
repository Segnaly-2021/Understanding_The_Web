## **Overview of Internet:**

We cannot think about the Web without talking about the Internet, this vast and complex network that connects computing devices - not just desktop PCs and servers but also connected things such as laptops, smartphones, tablets, TVs, watches, and more - across the world. In Internet jargon, all of these devices are called *hosts* or *end systems*. To communicate, one end system sends a message through the Internet, and the destination end system, maybe at the other side of the world, receives it.

Apart from end systems, the Internet is composed of **communication links** and **packet switches** - the hardware components - and a set of **protocols** that define the rules to be followed for a message to be succesfully sent through the internet. Internet is arguably a complex system, but to understand it better, let's use an analogy. 

The Internet is quite similar to the transportation networks of highways, roads and intersections. Just like the transportation networks allow vehicle to travel, the Internet allows data to travel from one end system to another. In many ways, communication links are analogous to roads, highways, packet switches are analogous to intersections, and protocols are analogous to traffic code.

From a software developer or an MLOps standpoint, knowing the Internet hardware components is quite enough as they rarely - or not at all - deal with them. That being said, let's now focus on the software part of the Internet, the protocols. And we have already seen one in this series, HTTP, in the last chapter. Remember that HTTP allows clients and Web servers to send messages to each other. Just like HTTP messages, any activities in tne Internet that involve two or more communicating remote entities are govern by protocol.

*A protocol defines the format and the order of messages exchanged between two or more communicating entities, as well as the actions taken on the transmission and/or the receipt of a message or other event.*

We as human use protocols (or at least good manner) all the time. Consider what you would if you ask someone for the time of the day. A typical discussion is shown in the figure below. It indicates you initiating the discussion by offering a greeting (the first 'Hi'). Then, if you get a positive answer (the second 'Hi' in the figure), you'll proceed by asking your question or if you get a negative answer, such as "Don't bother me!" or "Sorry, I don't speak english" or some unprintable reply :), the human protocol would be not to ask for the time of day. What's interesting to note is that there are specific messages we send, and specific actions we take depending on the received reply.

The internet has multiple protocols organized in layers called the Internet protocol stack.

## **The Internet Protocol Stack:**

The Internet protocol stack is structured in five layers, the application layer, the transport layer, the network layer, the link layer and finally the physical layer, each with its own set of protocols. Note, in the figure below, that not every protocol layer is implemented in every network component, especially for the lower level layers - the physical and link layers.As MLOps engineers, We'll focus on the top three protocol layers with their protocols as they are directly related to applications:

- **The application-layer (AL) protocols**:

  The AL is where network applications and their protocols such as **HTTP** (for web apps), **SMTP** (for e-mail apps), **FTP** (for file transport between end systems) reside. Network applications are the *raison d'être* of computer networking - if we couldn't conceive of any useful apps, there wouldn't be any need for networking infrastructure or protocols to support them. Applications use these protocols to exchange messages, we've seen the example of HTTP messages in the last chapter.
  
- **The Transport-layer (TL) protocols**:
  
    The TL provides two protocols, one with a connection-oriened service (TCP, or Transmission Control Protocol) and another with a connectionless-   oriented service (UDP, or User Datagram Protocol). These take application messages, potentially divide them into smaller chucks called segments and add additional headers to each segment.
  
- **The network-layer (NL) protocol**:

  The NL primarily uses the Internet Protocol (IP) to provide uniquely identifiable addresses, known as IP addresses, to any hosts and routers as shown in the figure below.
    
One of the greatest advantages of using the protocol-layering architecture is that it provides flexibility - notably, the ability to change the implementation of one layer without affecting others is of great importance - and it also abstracts the details of lower layers. Each layer provides its service by (1) performing certain actions within that layer and by (2) using the services of the layer directly below it. For example, from a web application perspective, HTTP seems to be the only protocol responsible for transporting messages; in reality HTTP is just a part of a larger system.

## **TCP/IP Connections:**

Now that we have a good understanding of what network protocols are, Let's focus on TCP/IP connections, as TCP is the transport-layer protocol used by HTTP.

As we'll see in the next chapter, a Web app is a pair of programs running on different end systems: a fixed, well-known-address Web server and a client. Servers host the web content, and clients can request it whenever they want. In the Operating System parlance, these programs are called processes. Processes send messages to each other through the computer network to communicate. Each program running on an end system or process has its own socket which binds it to the transport layer, as shown in the figure below. You can think of sockets as the doors of processes. For a process to send a message, it has to push it through its door. By doing this, the process assumes there is a transportation system on the other side of its door that will transports the message to the right socket (door) on the destination host. 

Because Operating system can handle multiple processes simultanously, you'll need to identify the right socket. The *port number* serves this purpose. 

TCP/IP connections are therefore uniquely identified by exactly 4 numbers:
 **\<source host IP address> \<source port number> : \<destination host IP address> \<destination port number>**
 
These connections can be either non-persistent or persistent. Non-persistent connections are closed just after a transaction (which consists of a request message and a response message) is completed. Since HTTP 1.1, the default mode of Web servers is persistent connections, where clients and servers can exchange multiple transactions. 

Aside from the connection-oriented service TCP/IP provides a reliable data transfer service. By adding a sequential number header and a error detection header, TCP/IP ensures that messages are delivered correctly and in order to the destination host.

**Note:** TCP/IP does not provide any security service, meaning messages being transmitted are not encrypted. The Internet developer community has developed a solution for this problem, namely the SSL (Secure Socket Layer). The SSL protocol is not a new protocol at the same level as TCP but rather an enhancement of TCP that should be implemented at the application layer.
